# <span style="color:blue">**Fundamentos  del machine  learning**</span>



# [Panorama general del aprendizaje automático](#)

El *machine learning* actualmente es clave en la ciencia de datos y en múltiples aplicaciones de uso cotidiano. Aunque su presencia masiva es relativamente reciente, sus fundamentos y primeras aplicaciones prácticas 

* como el reconocimiento óptico de caracteres y 
* los filtros de spam en los años noventa

demostraron su capacidad para aprender patrones a partir de datos y __automatizar decisiones__.

En la actualidad, el aprendizaje automático se encuentra en:
* los sistemas de búsqueda, 
* traducción automática, 
* reconocimiento de imágenes, 
* recomendaciones y asistentes inteligentes, 

impactando a miles de millones de usuarios. 
> Comprender qué significa que una máquina “aprenda” y en qué se diferencia del simple almacenamiento de información es esencial para su correcta aplicación en problemas reales.

## [  ¿Qué es el machine learning?](#) 

El **machine learning** es una disciplina central de la Ciencia de Datos que estudia cómo los sistemas computacionales pueden **aprender patrones y regularidades a partir de datos**, mejorando su desempeño en una tarea específica sin necesidad de ser programados explícitamente para cada caso.

Una definición clásica lo describe como:
<div class="alert alert-block alert-warning">
<b>[El machine learning es el]</b> campo de estudio que da a los ordenadores la capacidad de  aprender sin ser programados de manera explícita.  —Arthur Samuel, 1959 
</div>

Y una más orientada a la ingeniería:
<div class="alert alert-block alert-warning">
<b> </b> Se dice que un programa de ordenador aprende de la experiencia E, con respecto a una  tarea T y una medida de rendimiento R, si su rendimiento en T, medido por P, mejora  con la experiencia E.  —Tom Mitchell, 1997 
</div>


## Componentes clave del aprendizaje automático
Un sistema típico de *machine learning* se compone de los siguientes elementos:

- **Datos de entrenamiento (experiencia E):** Conjunto de ejemplos utilizados para aprender.
- **Instancias o muestras:** Cada ejemplo individual del conjunto de entrenamiento.
- **Modelo:** Representación matemática o computacional que captura patrones en los datos.
- **Tarea (T):** El problema que el sistema debe resolver (clasificación, regresión, predicción, etc.).
- **Medida de rendimiento (P):** Métrica cuantitativa para evaluar el desempeño del modelo  
  (por ejemplo, la exactitud).

Un ejemplo representativo es el **filtro de spam**, donde el modelo aprende a clasificar correos electrónicos como *spam* o *ham* a partir de ejemplos previamente etiquetados. 

* Spam → correo no deseado, publicidad masiva, fraudes, mensajes maliciosos
* Ham → correo válido, personal o profesional, legítimo

$\text{Correo} \in \{\text{spam}, \text{ham}\}$

>En este contexto, la **exactitud** es una métrica común para evaluar el desempeño del sistema.


 






## ¿Qué no es machine learning?

Es fundamental distinguir el aprendizaje automático del simple almacenamiento de datos. Disponer de grandes volúmenes de información (por ejemplo, una copia completa de Wikipedia) **no implica aprendizaje** si el sistema no mejora su desempeño en una tarea específica.

El *machine learning* requiere **aprendizaje a partir de la experiencia**, no únicamente la acumulación de datos.

### [¿Por qué utilizar machine learning?](#)

El *machine learning* resulta especialmente útil cuando los enfoques tradicionales basados en reglas se vuelven **complejos, frágiles o difíciles de mantener**, como ocurre al programar manualmente un filtro de spam (véase [Fig. 1.1](#f1-1)).

Un ejemplo es el reconocimiento de voz, ya que definir reglas manuales que cubran múltiples idiomas, hablantes, acentos y condiciones acústicas es inviable.

Finalmente, los modelos de *machine learning* no solo sirven para predecir, sino también también pueden **apoyarnos a la comprensión de los datos**, al revelar patrones ocultos y correlaciones inesperadas, un proceso estrechamente vinculado con la *minería de datos*.

- Reducir sistemas basados en reglas extensas y difíciles de mantener.
- Abordar problemas complejos sin soluciones algorítmicas claras.
- Adaptarse a entornos dinámicos mediante reentrenamiento.
- Extraer conocimiento a partir de grandes volúmenes de datos.

### Tipos de supervisión y algoritmos representativos

La clasificación de los sistemas de *machine learning* según el tipo de supervisión permite asociar
cada categoría con **familias de algoritmos bien definidos**, ampliamente utilizados en la práctica
y en investigación.


---

#### Aprendizaje supervisado
Utiliza conjuntos de datos etiquetados para aprender una función que mapea entradas a salidas.

**Algoritmos representativos:**
- Regresión lineal y regresión polinomial  
- Regresión logística  
- Máquinas de soporte vectorial (SVM / SVR)  
- k-vecinos más cercanos (k-NN)  



- Árboles de decisión  
- Random Forests  
- Gradient Boosting (XGBoost, LightGBM)  
- Redes neuronales artificiales (MLP, CNN, RNN, Transformers)

**Tareas típicas:** clasificación, regresión, predicción.

---

#### Aprendizaje no supervisado
Trabaja con datos sin etiquetas y busca estructuras, patrones o representaciones latentes.

**Algoritmos representativos:**
- k-means  
- DBSCAN  
- Clustering jerárquico  
- Modelos de mezclas gaussianas (GMM)  


- Análisis de componentes principales (PCA)  
- t-SNE, UMAP  
- Autoencoders

**Tareas típicas:** agrupamiento, reducción de dimensionalidad, detección de patrones.


---

#### Aprendizaje autosupervisado
Genera señales de supervisión a partir de los propios datos para aprender representaciones útiles.

**Algoritmos y enfoques representativos:**
- Autoencoders y autoencoders variacionales (VAE)  
- Contrastive Learning (SimCLR, MoCo)  
- Masked Language Models (BERT)  
- Masked Autoencoders (MAE)  
- Predictive Coding

**Tareas típicas:** preentrenamiento, aprendizaje de representaciones, transferencia de conocimiento.


---

#### Aprendizaje semisupervisado
Combina una pequeña cantidad de datos etiquetados con grandes volúmenes de datos no etiquetados.

**Algoritmos representativos:**
- Label propagation  
- Self-training  
- Co-training  
- Semi-supervised SVM  
- Pseudo-labeling con redes neuronales  
- Consistency regularization (FixMatch, Mean Teacher)

**Tareas típicas:** clasificación y regresión con etiquetas escasas.


---

#### Aprendizaje por refuerzo
Entrena agentes que aprenden a tomar decisiones mediante interacción con un entorno y recompensas.

**Algoritmos representativos:**
- Q-learning  
- SARSA  
- Deep Q-Networks (DQN)  
- Policy Gradient  
- Actor–Critic  
- Proximal Policy Optimization (PPO)  
- Deep Deterministic Policy Gradient (DDPG)

**Tareas típicas:** control, planificación, juegos, robótica.

---

En conjunto, estas categorías y algoritmos constituyen el **núcleo metodológico del machine learning**
y sirven como guía para seleccionar enfoques adecuados según la naturaleza de los datos, la
disponibilidad de etiquetas y los objetivos del problema.




# <span style="color:blue">**Aprendizaje supervisado**</span>



En el **aprendizaje supervisado**, el modelo se entrena con un conjunto de entrenamiento etiquetado, es decir, un conjunto de instancias de entrada para las cuales se conoce la salida correcta o etiqueta (véase Fig. 1.5).

En la figura, cada correo electrónico representa una instancia y su etiqueta indica si es spam o ham. El algoritmo utiliza estos ejemplos para aprender la relación entre las características del correo y su etiqueta.

Una vez entrenado, el modelo puede recibir una nueva instancia no etiquetada y predecir su clase, con la expectativa de que el conocimiento aprendido generalice correctamente a datos no vistos.

<div>
<img src="fig1-5.png" width="600"/>
</div>


Las tareas más comunes en este paradigma son:


#### Clasificación
Asigna una **clase discreta** a cada instancia.

- **Ejemplos generales:** filtro de spam, clasificación de noticias.  
- **Ejemplos científicos:**  
  - Identificación de partículas en detectores (electrón, muón, hadrón).  
  - Clasificación de eventos en rayos cósmicos por composición primaria.  
  - Diagnóstico médico asistido (benigno vs maligno).


**Métricas típicas:**
- Accuracy  
- Precision, Recall  
- F1-score  
- Curva ROC y AUC  
- Matriz de confusión  
- Tasa de falsos positivos / falsos negativos  

---

#### Regresión
Predice un **valor numérico continuo** a partir de un conjunto de características, como el precio de un coche o un parámetro físico experimental (véase Fig. 1.6).

<div class="alert alert-block alert-danger">
<b>Dato curioso </b> este nombre que suena tan raro es un término estadístico introducido por Francis Galton  mientras estudiaba el hecho de que los hijos de personas altas tienden a ser más bajos que sus progenitores.  Puesto que los hijos eran más bajos, llamó a esto "regresión a la media". Este nombre se aplicó después a  los métodos que utilizaba para analizar correlaciones entre variables.</div>

<div>
<img src="fig1-6.png" width="700"/>
</div>


- **Ejemplos generales:** predicción de precios, demanda energética.  
- **Ejemplos científicos:**  
  - Estimación de energía de partículas a partir de observables experimentales.  
  - Reconstrucción de parámetros físicos (energía, masa, Xmax) usando simulaciones Monte Carlo.  
  - Ajuste de curvas de luz o señales temporales en astrofísica y física experimental.





**Métricas típicas:**
- Error cuadrático medio (MSE)  
- Raíz del error cuadrático medio (RMSE)  
- Error absoluto medio (MAE)  
- Coeficiente de determinación \(R^2\)  
- Bias y resolución (muy comunes en física experimental)

---


Algunos modelos pueden emplearse tanto para regresión como para clasificación. Por ejemplo, la **regresión logística** se usa en clasificación al producir probabilidades de pertenencia a una clase.

<div class="alert alert-block alert-info">

> **Nota:** En aprendizaje supervisado, **etiqueta** y **objetivo** suelen usarse como sinónimos.  
> *Etiqueta* es más común en clasificación y *objetivo* en regresión.  
> Las variables de entrada también se denominan **características**, **predictores** o **atributos**.
</div>






En aplicaciones científicas, el aprendizaje supervisado es especialmente potente cuando las etiquetas provienen de **simulaciones Monte Carlo, modelos teóricos o datos calibrados**, permitiendo inferencia física robusta y cuantificación de incertidumbres.

# <span style="color:blue">**Aprendizaje autosupervisado**</span>



El aprendizaje autosupervisado crea etiquetas de manera automática a partir de datos no etiquetados. Por ejemplo, se pueden ocultar partes de una imagen y entrenar un modelo para reconstruir la versión original, usando la imagen incompleta como entrada y la imagen original como etiqueta. De este modo, los datos quedan efectivamente etiquetados y pueden emplearse posteriormente en algoritmos de aprendizaje supervisado (véase la [figura 1.12](#f1-12)).

<a id="f1-12"></a> 
<div>

El aprendizaje autosupervisado permite entrenar modelos útiles a partir de datos no etiquetados, como en la reparación de imágenes. Posteriormente, estos modelos pueden ajustarse a tareas específicas mediante un pequeño conjunto de datos etiquetados, por ejemplo, para clasificar especies de animales.

<div style="background-color: #d9f1ff; border-left: 10px solid #007acc; padding: 10px; border-radius: 5px; margin: 10px 0;">
    <strong>Nota:</strong> 
Transferir el conocimiento de una tarea a otra se denomina aprendizaje por transferencia  y es una de las técnicas del machine learning en la actualidad, sobre todo cuando se usan redes  neuronales profundas (es decir, redes neuronales compuestas por muchas capas de neuronas).
</div>

El aprendizaje autosupervisado, aunque trabaja con datos sin etiquetar, genera etiquetas durante el entrenamiento, acercándose al aprendizaje supervisado. Se enfoca en tareas como clasificación y regresión, a diferencia del aprendizaje no supervisado, que abarca agrupamiento y detección de anomalías. Por ello, es mejor considerarlo como una categoría independiente.

### [Aprendizaje por lotes frente a aprendizaje online](#)

Los sistemas de machine learning pueden clasificarse según si aprenden gradualmente a partir de datos entrantes o si el aprendizaje ocurre en bloques o lotes.

### [Aprendizaje por lotes](#)

En el aprendizaje por lotes, el sistema aprende utilizando todos los datos disponibles de una vez, lo que requiere tiempo y recursos, y generalmente se realiza offline. Una vez entrenado, el sistema se pone en producción y no sigue aprendiendo. Con el tiempo, el rendimiento puede disminuir debido a la evolución del mundo, lo que se conoce como "data drift" o deriva de datos. La solución es reentrenar el modelo regularmente con datos actualizados, dependiendo del caso de uso.


<div style="background-color: #e8ffe6; border-left: 10px solid #d9d9d9; padding: 10px; border-radius: 5px; margin: 10px 0;">
    <strong>Advertencia:</strong> 
Incluso un modelo entrenado para clasificar imágenes de gatos y perros puede  necesitar reentrenarse con regularidad, no porque los gatos y los perros muten de la noche a la  mañana, sino porque las cámaras siguen cambiando, junto con los formatos de imagen, la nitidez, el brillo y las proporciones. Además, puede que a la gente le encante una raza diferente el  año que viene o que decidan poner gorritos a sus mascotas, ¿quién sabe?. 
</div>


Si un sistema de aprendizaje por lotes necesita conocer datos nuevos, como un tipo nuevo de spam, debe ser reentrenado desde cero con todo el conjunto de datos (antiguos y nuevos). Aunque este proceso puede automatizarse, el reentrenamiento completo puede tardar horas, por lo que generalmente se realiza cada 24 horas o incluso semanalmente. Si los datos cambian rápidamente, se requiere una solución más reactiva. Además, entrenar con grandes cantidades de datos consume muchos recursos computacionales, lo que puede resultar costoso. En sistemas con recursos limitados, como en aplicaciones móviles o rovers en Marte, es más eficiente usar algoritmos que aprendan de manera gradual.

### [Aprendizaje online](#)

En el aprendizaje online, el modelo no se entrena de una sola vez con todos los datos, sino que se actualiza continuamente conforme llegan nuevas instancias, ya sea una por una o en minilotes. Cada actualización es computacionalmente ligera, lo que permite que el sistema aprenda en tiempo real, se adapte a cambios en los datos y mejore su desempeño incluso después de estar en producción (véase la [figura 1.14](#f1-14)).

El aprendizaje online es útil para sistemas que deben adaptarse rápidamente a cambios, como la detección de patrones en el mercado bursátil, o cuando los recursos computacionales son limitados, como en dispositivos móviles.

Además, se puede usar para entrenar sistemas en grandes conjuntos de datos que no caben en la memoria principal de una máquina, mediante un proceso conocido como "aprendizaje fuera del núcleo" (out of core). El algoritmo carga parte de los datos, entrena con ellos y repite el proceso hasta completar el entrenamiento con todo el conjunto de datos (véase la [figura 1.15](#f1-15)).

Un parámetro importante de los sistemas de aprendizaje online es lo rápido que deberían  adaptarse a los datos cambiantes: esto se denomina "tasa de aprendizaje". Si establecemos  una tasa de aprendizaje alta, entonces el sistema se adaptará con rapidez a datos nuevos, pero  también tenderá a olvidar rápido los datos antiguos (y no nos interesa que un filtro de spam  marque solo los últimos tipos de spam que se le han enseñado). Por el contrario, si establecemos  una tasa de aprendizaje baja, el sistema tendrá más inercia, es decir, aprenderá más despacio,  pero también será menos sensible al ruido de los datos nuevos o a las secuencias de puntos  de datos no representativos (valores atípicos). 



<div class="alert alert-block alert-info"><b>Advertencia:</b>    
El aprendizaje out of core suele hacerse offline (es decir, no en el sistema vivo), así  que el nombre "aprendizaje online" puede resultar confuso. Piense en ello como "aprendizaje  gradual". 
</div>

Un desafío del aprendizaje online es que la introducción de datos incorrectos puede reducir rápidamente el rendimiento del sistema. Esto puede suceder por fallos, como un sensor defectuoso en un robot, o por manipulación externa, como el spam en un motor de búsqueda. Para mitigar este riesgo, es crucial monitorear el sistema de cerca, desactivar el aprendizaje si detectamos una caída en el rendimiento y considerar revertir el sistema a un estado anterior. Además, se deben monitorear los datos de entrada y usar algoritmos de detección de anomalías para gestionar datos inusuales.

## <span style="color:green">**ENTRENAR EL MODELO**</span>

Antes de utilizar un modelo de regresión lineal para hacer predicciones, es necesario **determinar los valores adecuados de sus parámetros**, en este caso $ \theta_0 $ y $ \theta_1 $. Estos parámetros controlan la posición y la pendiente de la recta que el modelo utiliza para aproximar los datos.

Para decidir qué tan bueno es un conjunto dado de parámetros, necesitamos una **medida de rendimiento del modelo**. Existen dos enfoques equivalentes:

- Una **función de utilidad**, que mide qué tan bueno es el modelo (y que buscamos maximizar).
- Una **función de pérdida**, que mide qué tan malo es el modelo (y que buscamos minimizar).

En problemas de **regresión lineal**, es habitual trabajar con una **función de pérdida** que cuantifica la distancia entre las predicciones del modelo y los valores reales observados en el conjunto de entrenamiento. El objetivo es encontrar los valores de $ \theta_0 $ y $ \theta_1 $ que **minimizan dicha distancia**, logrando así el mejor ajuste posible a los datos.



El procedimiento automático mediante el cual se ajustan estos parámetros se conoce como **entrenamiento del modelo**. En este caso, al aplicar el algoritmo de regresión lineal, se obtiene que los valores óptimos de los parámetros son:

$$
\theta_0 = 3.75 \qquad \text{y} \qquad \theta_1 = 6.78 \times 10^{-5}.
$$

Estos valores definen la recta que mejor representa la relación entre las variables de entrada y salida según el criterio de ajuste elegido.

Veamos una manera rápida de obtener los valores anteriores usando `scikit-learn`

````python
from sklearn import linear_model

X_sample = country_stats[[gdppc_col]].values
y_sample = country_stats[[lifesat_col]].values

lin1 = linear_model.LinearRegression()
lin1.fit(X_sample, y_sample)

t0, t1 = lin1.intercept_[0], lin1.coef_[0][0]
print(f"θ_0={t0:.2f}, θ_1={t1:.2e}")
````

<b>Advertencia:</b><br><br>

Aunque puede resultar confuso al inicio, el término <b>modelo</b> se utiliza en aprendizaje automático con distintos significados, según el contexto:

- Puede referirse a un <b>tipo de modelo</b>, por ejemplo, una <i>regresión lineal</i>.
- También puede aludir a una <b>arquitectura de modelo completamente especificada</b>, como una regresión lineal con una sola variable de entrada y una sola variable de salida.
- Finalmente, puede significar el <b>modelo entrenado</b>, es decir, aquel que ya tiene valores concretos para sus parámetros y está listo para hacer predicciones, por ejemplo, una regresión lineal con una entrada y una salida, usando $ \theta_0 = 3.75 $ y $ \theta_1 = 6.78 \times 10^{-5} $.

La <b>selección de modelo</b> consiste en elegir el tipo de modelo y definir por completo su arquitectura.  
Por otro lado, <b>entrenar un modelo</b> significa ejecutar un algoritmo que ajuste los parámetros del modelo para que se adapte lo mejor posible a los datos de entrenamiento y, idealmente, sea capaz de realizar buenas predicciones sobre datos nuevos.
</div>

### Predicción con un modelo entrenado

El modelo de regresión lineal aprende una relación del tipo:

$$
\hat{y} = \theta_0 + \theta_1 x
$$

Para predecir la satisfacción con la vida de un país, se introduce su PIB per cápita como una **muestra con una sola característica**.

Aunque se trate de un único valor, el modelo espera una **entrada matricial**, ya que internamente opera con álgebra lineal.  
El resultado de la predicción es un arreglo de tamaño $(1 \times 1)$, del cual se extrae el valor numérico final.

>Por ejemplo,  supongamos que queremos saber lo felices que son los chipriotas y los datos de la OCDE no  tienen la respuesta.

<div class="alert alert-block alert-info">
<b>Regularización</b>  

Restringir un modelo para que sea más simple y reducir el riesgo de sobreajuste se llama <b>regularización</b>.  

- Un modelo lineal con dos parámetros ($\theta_0$ y $\theta_1$) tiene dos grados de libertad: altura y pendiente.  
- Si fijamos $\theta_1 = 0$, el modelo solo puede moverse hacia arriba o abajo, ajustándose alrededor de la media.  
- Permitir que $\theta_1$ cambie, pero manteniéndolo pequeño, da un modelo intermedio: más simple que el original, pero más flexible que el modelo plano.  

<b>Clave:</b> encontrar el equilibrio entre ajustar los datos y mantener el modelo lo suficientemente simple para generalizar.
</div>

<div class="alert alert-block alert-info">
<b>Regularización en la práctica</b>

  <li>Línea continua: modelo entrenado con todos los países (círculos + cuadrados), más representativo.</li>
  <li>Línea discontinua: modelo con regularización, pendiente más pequeña, ajusta peor los datos de entrenamiento, pero generaliza mejor a datos nuevos (cuadrados).</li>
</ul>

<b>Conclusión:</b> la regularización simplifica el modelo y mejora la generalización.

## Cómo solucionar el subajuste

- Seleccionar un modelo más potente, con más parámetros.  
- Añadir mejores características (**ingeniería de características**).  
- Reducir restricciones en el modelo (por ejemplo, disminuir la regularización).


## Retroceder: visión general del Machine Learning

- Machine Learning permite que las máquinas aprendan de datos, sin reglas explícitas.  
- Tipos de ML: supervisado / no supervisado, batch / online, basado en instancias / basado en modelos.  
- Algoritmo basado en modelos: ajusta parámetros para predecir bien en entrenamiento y nuevos casos.  
- Algoritmo basado en instancias: compara nuevas instancias con ejemplos almacenados usando similitud.


## Condiciones para un buen modelo

- Conjunto de entrenamiento **representativo**.  
- Datos limpios, con pocas características irrelevantes y sin demasiado ruido.  
- Modelo ni demasiado simple (subajuste) ni demasiado complejo (sobreajuste).

<div class="alert alert-block alert-info">
La única forma de saber si un modelo generaliza es probarlo con datos nuevos.  
- Dividir los datos en conjunto de entrenamiento y conjunto de prueba.  
- Error en datos de prueba = <b>error de generalización</b> o “fuera de muestra”.  
- Error bajo en entrenamiento pero alto en prueba → el modelo está <b>sobreajustando</b>.
</div>


<div class="alert alert-block alert-info"><b>Truco:</b>     
Es habitual utilizar el 80% de los datos para el entrenamiento y reservar el 20% para  las pruebas. Sin embargo, esto depende del tamaño del conjunto de datos: si contiene 10 millones de instancias, entonces reservar el 1% significa que el conjunto de prueba contendrá 100.000 instancias, probablemente más que suficientes para hacer una buena estimación  del error de generalización. 

    
</div>

## Evaluación y selección de modelo

- Evaluar un modelo se hace usando un **conjunto de prueba**.  
- Si tenemos varios modelos (por ejemplo, lineal vs polinomial), podemos compararlos usando el conjunto de prueba para ver cuál generaliza mejor.  
- Incluso si un modelo generaliza mejor, debemos ajustar hiperparámetros (por ejemplo, regularización) para evitar sobreajuste.

---

## Problema al ajustar hiperparámetros

- Entrenar muchos modelos con distintos hiperparámetros y escoger el mejor según el conjunto de prueba puede producir un **sobreajuste al conjunto de prueba**.  
- Esto significa que el modelo final puede no funcionar bien en datos nuevos.  

---

## Validación hold-out ([figura 1.25](#f1-25))

- Retener parte del conjunto de entrenamiento como **conjunto de validación (o dev)**.  
- Entrenar modelos candidatos en el resto del entrenamiento y evaluar en el conjunto de validación.  
- Elegir el modelo con mejor rendimiento en validación, luego entrenarlo en todo el conjunto de entrenamiento y validación para obtener el modelo final.

## Validación cruzada

- Si el conjunto de validación es muy pequeño, la evaluación puede ser imprecisa.  
- Si es demasiado grande, el entrenamiento del modelo final se hace con menos datos.  
- Solución: **validación cruzada**  
  - Dividir varias veces el conjunto de entrenamiento en entrenamiento y validación.  
  - Promediar el rendimiento de cada modelo.  
  - Ventaja: medida más precisa del rendimiento.  
  - Inconveniente: el tiempo de entrenamiento se multiplica.

## Discrepancia de datos

- A veces tenemos muchos datos de entrenamiento, pero **no representan los datos de producción**.  
- Ejemplo: entrenar una app de reconocimiento de flores usando millones de imágenes de la web, pero solo unas pocas son fotos tomadas con la app real.  
- Tanto el **conjunto de validación** como el de **prueba** deben ser representativos de los datos reales.  

---

## Problema de rendimiento

- Si el modelo falla en validación, no sabemos si es por:  
  1. Sobreajuste al conjunto de entrenamiento.  
  2. Discrepancia entre datos de entrenamiento y datos reales.  
- Necesitamos distinguir entre estas causas para tomar acciones correctivas.

---

## Solución: conjunto de entrenamiento-desarrollo ([figura 1.26](#f1-26))

- Retener parte del entrenamiento como **conjunto de entrenamiento-desarrollo**.  
- Evaluar el modelo en este conjunto:  
  - Si falla → sobreajuste → simplificar modelo, regularizar o limpiar/añadir datos.  
  - Si funciona bien → evaluar en conjunto dev (representativo de producción).

## Ajuste final

- Si el modelo falla en el conjunto dev → problema de **discrepancia de datos**.  
- Posibles soluciones: preprocesar los datos de entrenamiento para que se parezcan más a los datos reales y reentrenar.  
- Una vez que el modelo funcione bien en **entrenamiento-desarrollo** y **dev**, evaluarlo en **conjunto de prueba** para estimar rendimiento en producción.

<div class="alert alert-block alert-info"><b>Teorema "No hay almuerzo gratis"</b>     

Un modelo simplifica los datos para ignorar detalles irrelevantes y poder generalizar.  
Al elegir un modelo, hacemos **asunciones implícitas**: por ejemplo, un modelo lineal asume relaciones lineales y que las desviaciones son ruido.  

David Wolpert (1996, <a href="https://homl.info/8">link</a>) demostró que **no existe un modelo que funcione mejor para todos los datos**. Para algunos conjuntos, un modelo lineal será óptimo; para otros, una red neuronal. Por eso se llama **No Free Lunch (NFL)**.  
En la práctica, hacemos supuestos razonables y evaluamos solo un conjunto limitado de modelos adecuados.
</div>
